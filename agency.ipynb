```python
import json
from pathlib import Path
import numpy as np
import pandas as pd
import mne
from scipy.signal import hilbert
```

```python
config = json.loads(Path("data/config.json").read_text())

DATASET_PATH = Path(config["dataset_path"]).expanduser()
```

```python
all_ratings = []

for subject_dir in sorted(DATASET_PATH.glob("sub-*")):

    beh = subject_dir / "beh"

    tsv_files = list(beh.glob("*.tsv"))

    if not tsv_files:
        print(f"No TSV found in {beh}")
        continue

    tsv = tsv_files[0]

    ratings = pd.read_csv(tsv, sep="\t")

    ratings["subject"] = subject_dir.name

    all_ratings.append(ratings)

ratings = pd.concat(all_ratings, ignore_index=True)
```

```python
# Load dyad-level EEG data
trial_plv_df = pd.read_csv("plv_trial_results.csv")

# sanity check
print(ratings[["subject", "TrialNumber"]].dtypes)
print(trial_plv_df[["subject", "TrialNumber"]].dtypes)
```

```python
# Merge participant-level ratings with dyad-level EEG data
participant_df = ratings.merge(
    trial_plv_df,
    on=["subject", "TrialNumber"],
    how="inner",
    validate="many_to_one"
)

print(participant_df.shape)
print(participant_df.head())
```
